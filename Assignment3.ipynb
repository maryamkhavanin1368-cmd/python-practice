```python
def calculator(x,y,operation):
    if operation == "a":
        return x + y
    elif operation == "s":
        return x - y
    elif operation == "m":
        return x*y
    elif operation == "d":
        return x/y
    else:
        return "Invalid operation"
print(calculator(2, 5, "d"))
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length*self.width
r = Rectangle(5, 10)
print(r.area())
```
